This file will be responsible for creating alphbet set for the project. We are going with character level tokenization as of now, so will treat each character as one token.

```python
# this is the universal set for our project
alphabet = set()
```

```python
# adding the lowercase letters to the universal set
for unicode in range(ord('a'), ord('z') + 1):
    alphabet.add(chr(unicode))

# adding the uppercase letters to the universal set
for unicode in range(ord('A'), ord('Z') + 1):
    alphabet.add(chr(unicode))
    
for integer in range(0, 10):
    alphabet.add(str(integer))
```

```python
# adding brackets to the universal set
brackets = ['(', ')', '{', '}', '[', ']']
for bracket in brackets:
    alphabet.add(bracket)   
```

```python
#adding digits to the universal set
for unicode in range(ord('0'), ord('9') + 1):
    alphabet.add(chr(unicode))
```

```python
# adding special characters to the universal set
special_characters = ['!', '@', '#', '$', '%', '^', '&', '*', '+', '-', '=', '_', '~', '`', '|', '\\', ':', ';', '"', '<', '>', ',', '.', '?', '/']
for special_character in special_characters:
    alphabet.add(special_character) 
```

```python
# adding whitespace characters to the universal set
whitespace_characters = [' ', '\t', '\n', '\r', '\v', '\f']
for whitespace_character in whitespace_characters:
    alphabet.add(whitespace_character)
```

```python
print(alphabet)
```

```python
alphabet = sorted(alphabet)
```

## Final Version 1 alphabet

Refine the alphabet created above by removing unnecessary whitespace characters and adding the missing single quote used by Java character literals.

```python
# Refine the alphabet assembled in the cells above.
alphabet = set(alphabet)

alphabet.difference_update({"\r", "\v", "\f"})
alphabet.add("'")

alphabet = sorted(alphabet)
```

```python
# Each vocabulary entry must represent exactly one character.
assert all(len(character) == 1 for character in alphabet)
assert all(character in alphabet for character in ["'", '"', "\\", "_", " ", "\n", "\t"])

print(f"Vocabulary size: {len(alphabet)}")
print(alphabet)
```
